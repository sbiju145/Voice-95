```
import subprocess, time, re, os

env = os.environ.copy()
env["PORT"] = "3900"
env["OMNIVOICE_PORT"] = "3900"
env["OMNIVOICE_SERVER_MODE"] = "1"
env["OMNIVOICE_API_KEY"] = "3456"
env["OMNIVOICE_ALLOWED_ORIGINS"] = "https://voice-95.pages.dev"

server = subprocess.Popen(
    ["python3", "-m", "uvicorn", "backend.main:app", "--host", "0.0.0.0", "--port", "3900"],
    env=env
)
time.sleep(5)

tunnel = subprocess.Popen(
    ["cloudflared", "tunnel", "--url", "http://localhost:3900"],
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True
)

for line in tunnel.stdout:
    match = re.search(r"https://[a-zA-Z0-9-]+\.trycloudflare\.com", line)
    if match:
        print("\n✅ നിങ്ങളുടെ പുതിയ Backend URL ഇതാണ്:\n")
        print(match.group(0))
        print("\n👆 ഈ ലിങ്ക് കോപ്പി ചെയ്ത് voice-95.pages.dev-ലെ Settings-ൽ പേസ്റ്റ് ചെയ്യുക.")
        break
```
